```python
src = '../data/raw/HDFS.log'
dst = '../data/raw/HDFS_sample.log'

with open(src, errors='ignore') as fi, open(dst, 'w') as fo:
    for i, line in enumerate(fi):
        if i >= 200_000:
            break
        fo.write(line)

print("Sample created")
```
